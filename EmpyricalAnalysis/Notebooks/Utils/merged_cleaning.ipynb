## Specifically for Nauset strangeness where volume analyzed and runtime  values range way over 5 ml

```python
from pathlib import Path
import pandas as pd


def filter_runtime(source_dir, target_dir, max_runtime=1200):
    """
    Filter CSV files by removing rows where RunTime exceeds max_runtime.

    Parameters
    ----------
    source_dir : str or Path
        Directory containing input CSV files.
    target_dir : str or Path
        Directory where filtered files will be saved.
    max_runtime : float, optional
        Maximum allowed RunTime (default is 1200).
    """
    source_dir = Path(source_dir)
    target_dir = Path(target_dir)
    target_dir.mkdir(parents=True, exist_ok=True)

    for file in source_dir.glob("*.csv"):
        df = pd.read_csv(file)

        if "RunTime" not in df.columns:
            print(f"Skipping {file.name}: no 'RunTime' column.")
            continue

        original_rows = len(df)
        df_filtered = df[df["RunTime"] <= max_runtime]
        removed_rows = original_rows - len(df_filtered)

        output_file = target_dir / f"{file.stem}_filtered.csv"
        df_filtered.to_csv(output_file, index=False)

        print(
            f"{file.name}: removed {removed_rows} rows "
            f"({original_rows} -> {len(df_filtered)})"
        )
```
